# BƯỚC 3: THAO TÁC NÂNG CAO VỚI MẢNG NUMPY NDARRAY
## 3.1. Khởi tạo ma trận ngẫu nhiên 5x4

```python

import numpy as np

# Tạo ma trận ngẫu nhiên 5x4
np.random.seed(42)

matrix = np.random.randn(5, 4) * 10 + 50

print("Ma trận ban đầu:")
print(np.round(matrix, 2))
```

## 3.2. Cắt lát ma trận hai chiều (Slicing)

```python

# Lấy hàng có chỉ số 1, 2, 3
# Lấy cột có chỉ số 2, 3
sub_matrix = matrix[1:4, 2:4]

print("Ma trận con:")
print(np.round(sub_matrix, 2))
```

## 3.3. Tính toán các đại lượng thống kê theo cột

```python

# Tính thống kê theo từng cột
col_mean = np.mean(matrix, axis=0)
col_std = np.std(matrix, axis=0)
col_median = np.median(matrix, axis=0)

print("Trung bình từng cột:", np.round(col_mean, 2))
print("Độ lệch chuẩn từng cột:", np.round(col_std, 2))
print("Trung vị từng cột:", np.round(col_median, 2))
```

## 3.4. Broadcasting - Trừ trung bình từng cột

```python

# Trừ từng phần tử cho trung bình cột tương ứng
matrix_centered = matrix - col_mean

print("Ma trận sau Broadcasting:")
print(np.round(matrix_centered, 2))

# Kiểm tra trung bình từng cột sau biến đổi
print("Trung bình sau biến đổi:")
print(np.round(np.mean(matrix_centered, axis=0), 2))
```

# PHẦN IV: BÀI TẬP VẬN DỤNG VÀ MỞ RỘNG
# BÀI 1: XỬ LÝ MA TRẬN NUMPY 6x6
## 1.1. Khởi tạo ma trận ngẫu nhiên 6x6

```python

import numpy as np

# Cố định kết quả ngẫu nhiên
np.random.seed(42)

# Tạo ma trận 6x6 gồm số nguyên từ 10 đến 99
matrix6 = np.random.randint(10, 100, size=(6, 6))

print("Ma trận ban đầu:")
print(matrix6)
```

## 1.2. Câu a - Thay các phần tử đường chéo chính bằng 0

```python

# Thay các phần tử đường chéo chính bằng 0
np.fill_diagonal(matrix6, 0)

print("Ma trận sau khi thay đường chéo chính:")
print(matrix6)
```

## 1.3. Câu b - Lọc phần tử lớn hơn 50 và tính trung bình

```python

# Trích xuất các phần tử lớn hơn 50
greater_than_50 = matrix6[matrix6 > 50]

print("Các phần tử lớn hơn 50:")
print(greater_than_50)

# Tính trung bình cộng
average = np.mean(greater_than_50)

print("Trung bình cộng:", average)
```

## 1.4. Câu c - Chuẩn hóa ma trận bằng phương pháp Min-Max

```python

# Tìm giá trị nhỏ nhất và lớn nhất
min_val = np.min(matrix6)
max_val = np.max(matrix6)

# Chuẩn hóa Min-Max
if max_val != min_val:
    matrix_normalized = (
        (matrix6 - min_val) / (max_val - min_val)
    )
else:
    matrix_normalized = np.zeros_like(matrix6, dtype=float)

print("Giá trị nhỏ nhất:", min_val)
print("Giá trị lớn nhất:", max_val)

print("\nMa trận sau khi chuẩn hóa:")
print(np.round(matrix_normalized, 2))
```

## 1.5. Kiểm tra kết quả Bài tập 1

```python

print(" KIỂM TRA BÀI 1 ")

print("Kích thước ma trận:", matrix6.shape)

print("Đường chéo chính:", np.diag(matrix6))

print("Số phần tử lớn hơn 50:", len(greater_than_50))

print("Trung bình các phần tử > 50:", round(average, 2))

print("Min sau chuẩn hóa:", matrix_normalized.min())
print("Max sau chuẩn hóa:", matrix_normalized.max())
```

```python

# Lọc loài versicolor và sepal_length > 6.0
df_filtered = df_iris[
    (df_iris['species'] == 'versicolor') &
    (df_iris['sepal_length'] > 6.0)
]

print("Các hoa versicolor có sepal_length > 6.0:")
display(df_filtered)

print("Số bản ghi thỏa mãn:", len(df_filtered))
```

## 2.3. Câu b - Thêm cột Petal Ratio

```python

# Tạo cột petal_ratio
df_iris['petal_ratio'] = (
    df_iris['petal_length'] / df_iris['petal_width']
)

print("Dữ liệu Iris sau khi thêm cột petal_ratio:")
display(df_iris.head())

print("Kích thước dữ liệu mới:", df_iris.shape)
```

## 2.4. Câu c - Tính trung bình Petal Ratio theo từng loài

```python

# Nhóm dữ liệu theo loài hoa và tính trung bình
mean_ratio = df_iris.groupby('species')['petal_ratio'].mean()

print("Trung bình petal_ratio theo từng loài hoa:")
print(mean_ratio.round(2))
```

# BÀI 3: QUẢN LÝ ĐƠN HÀNG BẰNG PANDAS
## 3.1. Khởi tạo DataFrame gồm 10 đơn hàng

```python

import pandas as pd

# Tạo dữ liệu 10 đơn hàng
data_orders = {
    'MaDH': [
        'DH01', 'DH02', 'DH03', 'DH04', 'DH05',
        'DH06', 'DH07', 'DH08', 'DH09', 'DH10'
    ],
    'NgayDat': [
        '2026-10-01', '2026-10-02', '2026-10-03',
        '2026-10-04', '2026-10-05', '2026-10-06',
        '2026-10-07', '2026-10-08', '2026-10-09',
        '2026-10-10'
    ],
    'KhachHang': [
        'Nguyen Van An', 'Tran Thi Binh',
        'Le Van Cuong', 'Pham Thi Dung',
        'Hoang Van Em', 'Nguyen Thi Hoa',
        'Tran Van Nam', 'Le Thi Lan',
        'Pham Van Minh', 'Vo Thi Mai'
    ],
    'SoLuong': [2, 5, 1, 3, 4, 2, 6, 1, 5, 3],
    'DonGia': [
        1500000, 1200000, 4500000, 2000000,
        900000, 3500000, 800000, 6000000,
        1100000, 2500000
    ]
}

# Tạo DataFrame
df_orders = pd.DataFrame(data_orders)

# Chuyển cột ngày đặt sang kiểu datetime
df_orders['NgayDat'] = pd.to_datetime(df_orders['NgayDat'])

print("Danh sách 10 đơn hàng:")
display(df_orders)
```

## 3.2. Câu a - Tính thành tiền cho từng đơn hàng

```python

# Tính thành tiền cho từng đơn hàng
df_orders['ThanhTien'] = (
    df_orders['SoLuong'] * df_orders['DonGia']
)

print("Danh sách đơn hàng sau khi tính thành tiền:")
display(df_orders)
```

## 3.3. Câu b - Lọc đơn hàng có thành tiền trên 5 triệu đồng

```python

# Lọc đơn hàng có thành tiền lớn hơn 5 triệu
df_orders_large = df_orders[
    df_orders['ThanhTien'] > 5000000
]

print("Các đơn hàng có thành tiền > 5.000.000 VNĐ:")
display(df_orders_large)

print("Số đơn hàng thỏa mãn:", len(df_orders_large))
```
